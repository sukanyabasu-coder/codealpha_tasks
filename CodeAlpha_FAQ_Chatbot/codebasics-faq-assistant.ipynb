```python
# Importing important libraries

import pandas as pd
import numpy as np

# For text preprocessing
import re

# For converting text into vectors
from sklearn.feature_extraction.text import TfidfVectorizer

# For similarity checking
from sklearn.metrics.pairwise import cosine_similarity
```

```python
import os

print(os.listdir('/kaggle/input/datasets/sukanyabasuiem/codebasics-faq'))
```

```python
# Loading the dataset with encoding

df = pd.read_csv(
    '/kaggle/input/datasets/sukanyabasuiem/codebasics-faq/codebasics_faqs.csv',
    encoding='latin1'
)

# Display first 5 rows
df.head()
```

```python
# Display column names

print(df.columns)
```

```python
# Checking null values

df.isnull().sum()
```

```python
# Storing questions and answers separately

questions = df['prompt']
answers = df['response']
```

```python
# Function for cleaning text

def preprocess_text(text):
    
    # Convert text to lowercase
    text = text.lower()
    
    # Remove special characters and punctuation
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    
    return text
```

```python
# Applying preprocessing to all questions

cleaned_questions = questions.apply(preprocess_text)
```

```python
# Display cleaned questions

cleaned_questions.head()
```

```python
# Creating TF-IDF object

vectorizer = TfidfVectorizer(stop_words='english')
```

```python
# Transforming questions into numerical vectors

tfidf_matrix = vectorizer.fit_transform(cleaned_questions)
```

```python
# Checking matrix shape

print(tfidf_matrix.shape)
```

```python
# Chatbot function

def chatbot(user_input):
    
    # Clean the user input
    cleaned_input = preprocess_text(user_input)
    
    # Convert user input into vector
    user_vector = vectorizer.transform([cleaned_input])
    
    # Calculate similarity
    similarity = cosine_similarity(user_vector, tfidf_matrix)
    
    # Find most similar question
    index = similarity.argmax()
    
    # Get highest similarity score
    score = similarity[0][index]
    
    # If similarity is low
    if score < 0.3:
        return "Sorry, I don't understand the question."
    
    # Return best matching answer
    return answers.iloc[index]
```

```python
# Testing chatbot

question = " Why should I choose Codebasics?"

reply = chatbot(question)

print("User:", question)
print("Bot:", reply)
```

```python
# Interactive chatbot loop

print("Codebasics FAQ Chatbot")
print("Type 'exit' to stop the chatbot")

while True:
    
    # Taking user input
    user_question = input("\nYou: ")
    
    # Exit condition
    if user_question.lower() == 'exit':
        print("Bot: Goodbye!")
        break
    
    # Get chatbot response
    bot_reply = chatbot(user_question)
    
    # Print response
    print("Bot:", bot_reply)
```
