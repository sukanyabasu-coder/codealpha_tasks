```python
from deep_translator import GoogleTranslator
from langdetect import detect

# Supported languages
languages = [
    "english",
    "hindi",
    "french",
    "german",
    "spanish",
    "japanese",
    "korean",
    "italian"
]

while True:

    try:
        print("\n===== Language Translator =====")

        # Display supported languages
        print("\nSupported Languages:")

        for lang in languages:
            print("-", lang)

        # Taking text input
        text = input("\nEnter text to translate: ")

        # Empty input check
        if text == "":
            print("Please enter some text.")
            continue

        # Detecting language
        detected_language = detect(text)

        print("\nDetected Language Code:", detected_language)

        # Taking target language
        target_lang = input("Enter target language: ").lower()

        # Checking valid language
        if target_lang not in languages:
            print("Language not supported.")
            continue

        # Translating text
        translated = GoogleTranslator(
            source='auto',
            target=target_lang
        ).translate(text)

        # Displaying output
        print("\nOriginal Text:")
        print(text)

        print("\nTranslated Text:")
        print(translated)

    except Exception as e:
        print("\nError occurred.")
        print(e)

    # Asking user to continue
    choice = input("\nTranslate again? (yes/no): ").lower()

    # Exit condition
    if choice == "no":
        print("\nProgram Ended.")
        break
```
